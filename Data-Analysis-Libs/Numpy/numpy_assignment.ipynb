````python
"""
1. Create a 3x3 array with random integers between 1 and 100. Write a program to
"normalize" the array so that the values are scaled between 0 and 1.
• Formula: (x - min) / (max - min)
• Example Output: An array where the smallest value becomes 0.0 and the
largest becomes 1.0.
"""
def normalize_arr(arr):
    min=arr.min()
    max=arr.max()
    for index,i in enumerate(arr):
        i=(i-min)/(max-min)
        arr[index]=i
arr=np.random.randint(1,101,size=(3,3)).astype(np.float32)
# converted to float to avoid all numbers being 0/1
normalize_arr(arr)
print(f"Normalized array:\n {arr}")
````

````python
"""
Create a 1D array of 10 random numbers. Ask the user for a value X Find the value
in the array that is closest to X
• Example Input: Array: [1, 5, 8, 12], X = 10
• Example Output: 8 or 12 (whichever is mathematically closer)
"""
def solution(arr,x):
    ans=arr[0] # assuming that 1st number is closest
    diff=abs(x-ans) 
    for i in arr:
        new_diff=abs(x-i)
        if new_diff<= diff: # closer
            ans=i
            diff=new_diff
    return ans
arr=np.random.randint(1,101,10)
print(arr)
x=int(input("Enter a value:"))
print(solution(arr,x))
````

````python
"""
Write a NumPy program to create a 5x5 array with random values. For each column
in the array, find the second-largest value.
• Example Input: [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
• Example Output: [4, 5, 6] (These are the second-highest values for each
respective column).
"""
def solution(arr):
    sorted_arr=np.sort(arr,axis=0) # sorting all columns
    print(arr)
    print(sorted_arr)
    second_largest=sorted_arr[3,:]
    return second_largest
arr=np.random.randint(1,101,size=(5,5))

print(solution(arr))
````

````python
"""
4. Create a 6x4 matrix of random integers. Identify which rows have a sum that is an
even number. Print the indices of these rows.
"""
def solution(arr):
    for i in range(0,arr.shape[0]):
        total=arr[i,:].sum() # calc sum for each row
        if total % 2 == 0: # even
            print(f"Sum of row with index {i} is {total}")
arr=np.random.randint(1,101,size=(6,4))
print(arr)

solution(arr)
````

````python
"""
5. Create a 1D array of 30 random integers between 1 and 50. Replace all numbers
that are multiples of 3 or 5 with the value -1.
• Example Input: [3, 7, 10, 15]
• Example Output: [-1, 7, -1, -1]
"""
def solution(arr):
    for index,i in enumerate(arr):
        if i%3==0 or i%5==0: # multiple
            arr[index]=-1
            
arr=np.random.randint(1,51,30)
print(arr)
solution(arr)
print(arr)
````

````python
"""
6. Write a NumPy program to create a 3x3 array with random values. For each row,
calculate its mean (average) and subtract that mean from every element in that
specific row.
• Example Input: ``` [[10, 20, 30], [5, 5, 5], [0, 10, 20]]
• Example Output: [[-10. 0. 10.], [ 0. 0. 0.], [-10. 0. 10.]]

"""
def solution(arr):
    for i in range(0,arr.shape[0]):
        mean=arr[i,:].mean() # calc mean for each row
        arr[i,:]-=mean
arr=np.random.randint(1,101,size=(3,3)).astype(np.float32)
print(f"Before:\n {arr}")
solution(arr)
print(f"After:\n {arr}")
````
