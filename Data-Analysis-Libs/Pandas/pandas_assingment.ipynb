```python

import pandas as pd
import numpy as np
```

```python
"""
1. Data Loading & Inspection Load the company_data.csv file. Print the first 10
rows and use a Pandas method to show the data types and check for missing
values in each column."""



df = pd.read_csv("company_data.csv")
print(df.head(10))
print(df.dtypes)
print(df.isnull().sum())
```

```python
"""
2. Handling Missing Values The dataset contains missing values in the Salary and
Bonus columns.
• Fill the missing Salary values with the median salary of the entire
company.
• Fill the missing Bonus values with 0."""

df["Salary"] = df["Salary"].fillna(df["Salary"].median())
df["Bonus"] = df["Bonus"].fillna(0)
```

```python
"""
3. Data Cleaning: The Employee_Name column is in the format "FirstName
LastName". Split this column into two new columns: First_Name and Last_Name,
then delete the original Employee_Name column."""

df[['First_Name', 'Last_Name']] = df['Employee_Name'].str.split(n=1, expand=True)
df.drop(columns='Employee_Name', inplace=True)
```

```python
"""
4. Filtering :Create a new DataFrame called Senior_IT that contains only employees
from the 'IT' department who have more than 8 years of experience."""

Senior_IT = df[(df['Department'] == 'IT') & (df['Years_Experience'] > 8)]
Senior_IT
```

```python
"""
5. Performance Mapping Create a new column called Performance_Level based on
the Evaluation_Score (1-100):
• Score > 80: "High"
• Score 50-80: "Medium"
• Score < 50: "Low"""

df["Performance_Level"] = np.where(df["Evaluation_Score"] > 80,"High",
    np.where(df["Evaluation_Score"] >= 50,"Medium","Low")
)
```

```python
"""
6. Calculate the total salary expense and the average evaluation score for each
Department."""
department_summary = df.groupby('Department').agg(
    Total_Salary=('Salary', 'sum'),
    Average_Evaluation_Score=('Evaluation_Score', 'mean')
)
department_summary
```

```python
"""
7. Find the top 5 highest-paid employees in the company. Print only their names
and salaries.
"""
top_5 = df.nlargest(5, 'Salary')[['First_Name', 'Last_Name', 'Salary']]
print(top_5.to_string(index=False))
```

```python
"""
8. The company is giving a 10% raise to everyone in the 'Sales' department. Update
the Salary column for those employees only.
"""
sales_mask = df["Department"] == "Sales"
df.loc[sales_mask, "Salary"] = (df.loc[sales_mask, "Salary"] * 1.10).round(2)
```

```python
"""9. Exporting Results Save the final, cleaned, and updated DataFrame to a new Excel
or CSV file named final_employee_report.csv."""
df.to_csv("final_csv", index=False)
```
