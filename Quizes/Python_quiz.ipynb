## Q91: Python functions

Write summarize_numbers(values) for a non-empty list of numbers. Return a dictionary with keys count, mean, min, and max. Raise ValueError for an empty list. Do not use NumPy or pandas. Demonstrate it on [2, 4, 6].

**Grading (5 marks):** Count and mean: 2; Minimum and maximum: 2; Empty-input handling and demonstration: 1

```python
def summarize_numbers(values):
    if values is None:
        raise ValueError
    else:
        summ_dict= {}
        summ_dict["count"] = len(values)
        total = 0
        for i in values:
            total += i
        summ_dict["sum"] = total
        summ_dict["mean"] = total / len(values)
        summ_dict["min"] = min(values)
        summ_dict["max"] = max(values)
        return summ_dict
            
my_list = [2, 4, 6]
dict = summarize_numbers(my_list)
print(dict)
```

**Your explanation:**
check if list is None so empty will raise ValueError
else will compute each of the requirements

## Q92: Python filtering

Write unique_even_squares(values). Return the sorted unique squares of even integers in the input list. Demonstrate it on [3, 2, 2, -4, 5, 0]. Do not use NumPy or pandas.

**Grading (5 marks):** Even filtering: 2; Squaring and deduplication: 2; Sorted result and demonstration: 1

```python
def unique_even_squares(values):
    set_values = set(values) # remove duplicates
    even_squares = {i**2 for i in set_values if i % 2 == 0}
    return list(even_squares)

my_list = [3, 2, 2, -4, 5, 0]
unique_even_squares(my_list)
```

**Your explanation:**
Used sets as they auto remove dupes,
then looped through each value in set, checked if even then squared the number and put into a new set (filtering if even)
returned final set as list to convert it to it's original type

## Q93: NumPy standardization

Using NumPy, standardize each column of X with the population standard deviation (ddof=0). Preserve the shape. Set standardized values of constant columns to zero. Store the result in X_scaled. This is a standalone numeric exercise, not a train/test modeling exercise.

**Grading (5 marks):** Columnwise mean/std: 2; Correct standardization: 2; Constant-column handling: 1

```python
X = np.array([[1., 10., 5.], [2., 20., 5.], [3., 30., 5.]])

mean = np.mean(X, axis=0) # calculate mean along each column
std = np.std(X, axis=0, ddof=0) # calculate standard deviation along each column with population formula (ddof=0)

X_scaled = np.zeros_like(X) 
# Use np.divide to scale the data, handling division by zero to prevent nan
np.divide(X - mean, std, out=X_scaled, where=(std != 0))

print(X_scaled)
```

**Your explanation:**
Calculated mean and std for each column
then created an arrays of zeros, so 0 becomes default if any calculation failed.
then calculated X_scaled using (x-mean)/std and stored it in X_scaled, used where to prevent any calc if std = 0
