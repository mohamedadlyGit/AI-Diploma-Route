```python
import seaborn as sns
# Advanced data visualization
```

```python
"""
 Seaborn is a high-level library that is built on top of Matplotlib. It provides a simpler interface for creating common types of statistical plots,
 such as scatter plots, line plots, and bar plots.
 Seaborn also provides more advanced statistical visualizations,
 such as violin plots, box plots, and heatmaps.
 Seaborn is designed to work well with Pandas data frames, 
 and it provides built-in support for many common data visualization tasks, 
 such as grouping data by categories and computing summary statistics.


In general, if you need to create complex visualizations or have very specific requirements for your plots,
Matplotlib may be the better choice. If you want to create common types of statistical plots quickly and easily,
or if you are working with Pandas data frames, 
Seaborn may be the better choice.
"""
```

```python
print(sns.get_dataset_names())
```

```python
df=sns.load_dataset("tips")
print(df.head())
print(df.info())
```

```python
# Scatter plot
sns.scatterplot(data=df, x="total_bill", y="tip",hue="sex")
# hue uses different colors to separate data into categories
```

```python
sns.scatterplot(data=df, x="total_bill", y="tip",hue="time")
```

```python
# bar plot
sns.barplot(data=df, x="day", y="total_bill",hue="sex")
```

```python
# Countplot
sns.countplot(data=df,x="day",hue="sex")
```

```python
# Box-Whisker plot
sns.boxplot(data=df, x="day", y="total_bill",hue="time")
```

```python
# Histogram plot
sns.histplot(data=df, x="total_bill",hue="sex")
```

```python
sns.histplot(data=df, x="total_bill",hue="sex", kde=True)
# - The KDE curve shows the estimated shape of the data distribution.
```

```python
sns.histplot(x=df["total_bill"],hue=df["sex"])
# without passing data=
```

```python
sns.pairplot(data=df,hue="sex")
# creates multiple plots showing the relationships between every pair of numerical columns in df.
# takes no x,y args
```

```python
sns.pairplot(data=df,hue="sex",
    kind="reg",       # Regression plots instead of scatter plots
    diag_kind="hist"  # Histograms on the diagonal
)
```

```python
corr = df.select_dtypes("number").corr()
#calculates the correlation between every pair of numerical columns in a DataFrame.
corr
```

```python
df.corr(numeric_only=True) # same as above
```

```python
df.corr(numeric_only=True)["total_bill"]
```

```python
sns.heatmap(corr,annot=True)
# annot means annotation.
```
