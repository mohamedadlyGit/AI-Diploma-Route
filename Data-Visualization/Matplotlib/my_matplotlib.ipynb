```python
# bar-chart
a = np.array(["class1", "class2", "class3"])
b = np.array([22,31,29])

plt.bar(a,b)
plt.xlabel("CLASS")
plt.ylabel("AMOUNT")
plt.show()
```

```python
# horizontal bar (switch x & y)
a = np.array(["class1", "class2", "class3"])
b = np.array([22,31,29])

plt.barh(a,b,color="orange")
plt.xlabel("AMOUNT")
plt.ylabel("CLASS")
plt.show()
```

```python
# pie chart
z = np.array([25,50,60])
labels = np.array(["Java","Python","C++"])
plt.pie(z, labels=labels)
plt.show()
```

```python
plt.subplot(2,2,1) # rows, columns, panel/positions
# max panels is rows x columns
plt.plot(x,x)
plt.title("Line-Plot")
plt.xlabel("x-label")
plt.ylabel("y-label")

plt.subplot(2,2,2)
plt.plot(x,np.sin(x))
plt.title("Line-Plot")
plt.xlabel("x-label")
plt.ylabel("y-label")

plt.subplot(2,2,3)
plt.plot(x,np.cos(x))
plt.title("Line-Plot")
plt.xlabel("x-label")
plt.ylabel("y-label")

plt.subplot(2,2,4)
plt.plot(x,np.tan(x))
plt.title("Line-Plot")
plt.xlabel("x-label")
plt.ylabel("y-label")

plt.tight_layout() #automatically adjusts spacing between plot elements so they don’t overlap or get cut off.
plt.show()
```

```python
"""
subplot() → singular; creates/selects one plot at a time.
subplots() → plural; creates the whole layout at once.
subplots() is generally cleaner and preferred for multiple plots.
when using subplots ,add a another var axes and any attribute use "set_attribute"
except PLOT no need
"""
```

```python
fig = plt.figure()

ax1 = fig.add_subplot(2,2,1)
ax1.plot(x,np.cos(x))

ax2 = fig.add_subplot(2,2,2)
ax2.plot(x,np.sin(x))

ax3 = fig.add_subplot(2,2,3)
ax3.plot(x,np.tan(x))

ax4 = fig.add_subplot(2,2,4)
ax4.plot(x,np.tanh(x))
```
